## Background

**Dissolved oxygen** (DO) is the amount of oxygen gas dissolved in water, reported in **milligrams of O₂ per liter** (mg/L). Fish breathe this oxygen through their gills, the same way you breathe oxygen from air through your lungs — if the concentration falls too low, they suffocate. A trout needs roughly 6 mg/L to thrive; a catfish can tolerate around 3 mg/L; below about 2 mg/L (a condition called **hypoxia**) most fish die.

DO varies on a **daily cycle** in any stream with algae or aquatic plants. During daylight, photosynthesis adds oxygen to the water and DO rises; at night, the same plants (plus bacteria and animals) consume oxygen through respiration, and DO falls. The lowest DO of the day is therefore almost always in the pre-dawn hours, just before sunrise — and that is when fish kills, if they happen, tend to happen. A monitoring station that only sampled at noon would miss the entire problem.

> For this assignment, treat the **regulatory threshold** as 6.0 mg/L (a typical value for a warm-water fishery in U.S. state water-quality regulations). Any hourly reading below this counts as a violation.

No calculus is needed — every quantity in this assignment is either a sum, a count, a minimum, or a maximum over a fixed list of 24 numbers.

## Your Solution

Write your code in the cell below. You may add additional cells (code or markdown) as needed to organize your work or explain your thinking.

````python
# Your solution here

do_readings = [
    7.8, 7.6, 7.5, 7.4, 7.3, 7.2,    # 00:00 - 05:00
    7.4, 7.7, 8.1, 8.6, 9.2, 9.7,    # 06:00 - 11:00
    10.1, 10.3, 10.2, 9.8, 9.2, 8.4, # 12:00 - 17:00
    7.6, 6.9, 6.3, 5.8, 5.5, 5.4,    # 18:00 - 23:00
]

THRESHOLD = 6.0  # mg/L

print("Stream Dissolved Oxygen Report - 24-hour record\nThreshold for aquatic life: 6.0 mg/L\n")

sumR = 0
maxR = 0
minR = 10
ind = 0

hr = 0
#loop for max min and sum of all
for i in do_readings:
    sumR += i
    if maxR == 0 or i >= maxR:
        maxR = i
    if i <= minR:
        minR = i
    ind +=1

    if i < THRESHOLD:
        hr += 1

#gets indexes/hours of min and max
iMin = do_readings.index(minR)
iMax = do_readings.index(maxR)


print(f"Daily Minimum: {minR:8.2f} mg/L at hour {iMin:02d}:00")
print(f"Daily Maximum: {maxR:8.2f} mg/L at hour {iMax:02d}:00 ")
print(f"Daily Mean: {sumR/ind:11.2f} mg/L")

#split for overnight mean
part = do_readings[0:6]

print(f"Overnight mean (00:00-05:00): {sum(part) / 6:.2f} mg/L")


print(f"\nHours below threshold: {hr} of 24")
if hr == 0:
    print("Status: PASS")
else:
    print(f"Status: FAIL - stream was below {THRESHOLD} mg/L for {hr} hour(s)")
````
