## Background

**Newton's method** (sometimes called the *Newton–Raphson* method) is a general technique for finding the roots of a function — that is, the values of $x$ where some function $f(x) = 0$. The idea is to start with a guess and repeatedly improve it by following the tangent line of $f$ down to the $x$-axis. You may meet a full derivation later in calculus; for now, all you need is the formula for the specific case of square roots.

Finding $\sqrt{S}$ is the same as finding the positive root of $f(x) = x^2 - S$. Newton's update rule reduces, in this case, to a beautifully simple iteration known to the Babylonians around 1700 BCE:

> Given a positive number $S$ whose square root you want, and a current guess $x_n$, the next (better) guess is:
>
> $$x_{n+1} = \tfrac{1}{2}\left( x_n + \frac{S}{x_n} \right)$$
>
> Intuition: if $x_n$ is too large then $S / x_n$ is too small, and averaging them gives something in between — closer to the true $\sqrt{S}$. The procedure roughly *doubles the number of correct digits* with every iteration, which is extremely fast.

**Stopping the loop.** You stop iterating when consecutive guesses are close enough to each other that further work is pointless. Define a small positive number $\varepsilon$ (epsilon) called the **tolerance**, and stop when:

$$|x_{n+1} - x_n| < \varepsilon$$

where the vertical bars mean *absolute value* (Python's built-in `abs()`).

**Why you cannot test for exact equality.** Floating-point numbers in a computer have finite precision: `0.1 + 0.2` famously does not equal exactly `0.3`. Two iterates of Newton's method will essentially never be bit-for-bit equal, so a test like `while x_new != x_old` would loop forever. Using a tolerance is the standard fix.

Write a Python program that asks the user for a non-negative number $S$ and computes $\sqrt{S}$ using Newton's iterative method. ## Your Solution

Write your code in the cell below. You may add additional cells (code or markdown) as needed to organize your work or explain your thinking.

````python
# Your solution here
import math

S = float(input("Enter a non-negative number: "))


e = 1e-10 # e for tolerance


def nSQ (s, t = e):
    i = 1 #indexing for iterations
    
    if s == 0:
        return 0.0, 0
    
    xN = s/2.0 # init guess
    x_next = 0.5 * (xN + s / xN) #next x guess
    print(f"Iteration {i}: {x_next:.10f}")
    #3 lines above first iteration to set varaibles up to be used correctly in the while loop below
    
    while (abs(x_next - xN) > t): #checks to see if close enough together for tolerance
        i += 1
        
        xN = x_next
        x_next = (1/2) * (xN + (s / xN))
        print(f"Iteration {i}: {x_next:.10f}")

    return x_next, i #when close enough returns

if S >= 0:
    k,j = nSQ(S)
    print(f"\nNewtons method result: {k:.10f}")
    print(f"math.sqrt({S})       : {math.sqrt(S):.10f}")
    print(f"Absolute difference: {math.sqrt(S) - k:.10f} ")
    
    print(f"Iterations Used: {j}")
else:
    print(f"Error: {S} is negative. The square root of a negative number is not real")

````
